```python
import pandas as pd

import warnings
warnings.filterwarnings('ignore')
```

```python
data = pd.read_csv('../data/raw/housing.csv')

print(data.head())
print(data.info())
```

```python
# fill up missing values
for col in data.columns:
    if data[col].isnull().sum() > 0:
        data[col].fillna(data[col].median(), inplace=True)
```
